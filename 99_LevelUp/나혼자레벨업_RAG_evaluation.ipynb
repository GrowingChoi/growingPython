## Retriver

- Query(비정형)을 입력 같아서 Vector store에서 관련된 내용을 검색하는 기능을 제공하는 인터페이스로 다양한 데이터소스에서 정보를 검색하여 LLM 기반 어플의 정확성을 향상시키는데 핵심 역할을 함

| Retriever 종류                          | 핵심 개념                                                   | 장점                        | 활용 예시                 |
| ------------------------------------- | ------------------------------------------------------- | ------------------------- | --------------------- |
| **1. Vector Store Retriever**         | 텍스트 청크를 임베딩 후 벡터 공간에 저장 → 쿼리 임베딩과 **코사인 유사도**로 유사 문서 검색 | 구현이 간단하고 빠름               | 기본 검색 시스템             |
| **2. ParentDocumentRetriever**        | 청크별 인덱싱 후 → 유사한 청크의 **원본 문서 전체** 반환                     | 문맥 유지에 유리                 | 논문, 리포트 등 문서 단위 검색    |
| **3. MultiVectorRetriever**           | 요약, 질의 생성 등으로 **문서당 다중 벡터 생성**                          | 문서 핵심 정보만 반영 가능           | 긴 문서, 핵심 문단 집중 검색     |
| **4. SelfQueryRetriever**             | **LLM을 이용해** 질문을 메타데이터 필터 + 검색어로 변환                     | 메타데이터 기반 고급 검색 가능         | "2023년에 작성된 리포트만 보여줘" |
| **5. ContextualCompressionRetriever** | 기존 검색 결과에서 불필요 정보 제거 → **압축 요약된 문서 반환**                 | LLM 입력 토큰 수 절약, 정밀한 요약 제공 | 요약 중심 질의 응답           |
| **6. MultiQueryRetriever**            | 쿼리를 LLM이 다양한 형태로 재작성 → 다각도 검색 수행                        | 다양한 표현에 강함, 검색 누락 방지      | 유연한 사용자 질문 대응         |
| **7. EnsembleRetriever**              | 여러 Retriever (BM25, 벡터 등)를 결합 → **가중치 기반 재정렬**          | 다양한 방식의 장점 통합             | 최종 성능 극대화 목적          |
