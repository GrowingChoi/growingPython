```python
question_texts = df['Q']
answer_texts = df['A']
all_texts = list(question_texts + " "+ answer_texts)
len(question_texts), len(answer_texts), len(all_texts)
```

```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.pre_tokenizers import Whitespace
from tokenizers.trainers import BpeTrainer

vocab_size = 10_000
min_frequency = 5

tokenizer = Tokenizer(BPE(unk_token="[UNK]"))
tokenizer.pre_tokenizer = Whitespace()
trainer = BpeTrainer(
    vocab_size = vocab_size
    , min_frequency = min_frequency
    , continuing_subword_prefix = '##'
    , special_tokens =["[PAD]","[UNK]","[SOS]","[EOS]"]
)
tokenizer.train_from_iterator(all_texts, trainer=trainer)


```

```python
#코드 토큰화 

input_text = "나는 최근에 파리 여행을 다녀왔다"
input_text_list = input_text.split()
print(input_text_list) 
```

```python
str2idx = {word:idx for idx, word in enumerate(input_text_list)}
idx2str = {idx:word for idx, word in enumerate(input_text_list)}

print('str2idx : ', str2idx)
print('idx2str : ', idx2str)

#토큰을 토큰 아이디로 변환
input_ids = [str2idx[word] for word in input_text_list]
print('input_ids : ', input_ids)
```

```python
import torch
import torch.nn as nn


embedding_dim = 16
embed_layer = nn.Embedding(len(str2idx),embedding_dim)
input_embeddings = embed_layer(torch.tensor(input_ids))
input_embeddings = input_embeddings.unsqueeze(0)
input_embeddings.shape
```

```python
import os
import glob
import json

# 디렉토리 설정
faq_dir = "csTemplate/"
md_files = glob.glob(os.path.join(faq_dir, "*.md"))

# 변환된 FAQ 리스트
faq_list = []

# 변환 로직
for filepath in md_files:
    print('이거 돌고잇음?')
    with open(filepath, "r", encoding="utf-8") as f:
        content = f.read()
    
    filename = os.path.basename(filepath)
    title = os.path.splitext(filename)[0]
    
    # 간단한 전처리: 텍스트 이어붙이기, 공백 제거
    body = content.replace("\n", " ").strip()
    
    faq_list.append({
        "question": f"{title} 관련 문의",
        "answer": body
    })

# 저장할 JSON 파일 경로
output_path = "data/converted_faq_data.json"
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(faq_list, f, ensure_ascii=False, indent=2)

output_path
```
