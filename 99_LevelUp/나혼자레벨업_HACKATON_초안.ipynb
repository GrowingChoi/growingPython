## 5. 아이디어 소개

### 1. 추진배경
지방 발전을 위한 예산은 다양한 사업에 투자되고 있지만, 일부는 사업이익의 최소치에 도달하지 못하거나, 비효율적으로 사용되는 사례가 존재함.(이건 검색해서 알아봐야할 내용 ㅇㅇ). 특히 인구 감소, 저출산 문제 해결 등 시급한 과제들에 자원이 집중되기 위해선 최소이익에 도달하지 못하는 재정의 절약이 우선순위가 되어야한다고 생각했습니다. 그래서 저희는 신뢰할 수 있는 공공 데이터(통계청, 열린재정 등)를 기반으로 지방 재정 예산을 효율적으로 사용할 수 있는 도구를 개발해야될 필요성을 느꼈습니다.


### 2. 목표 및 내용  
- **목표**: 공공 데이터를 기반으로 지방 예산의 낭비를 분석하고, 절약 가능성을 시각화하여 공무원이 효율적으로 예산을 재편할 수 있도록 돕는 App 개발
- **내용**:
  - 지방 예산 사용 데이터를 수집하고, ML/DL 모델을 통해 이상 지출 또는 낭비 가능성이 높은 항목을 탐지
  - 각 사업에 대한 투자 대비 효율을 분석하여 대안 제시
  - 시각화 대시보드로 지역/사업별 집행 구조를 직관적으로 제공
  - 사용자가 시뮬레이션을 통해 예산 재배분 결과를 시각화하여 제공

### 4. 기대효과 및 활용 방안  
- **공무원 대상**: 신뢰성있는 데이터의 AI분석을 통해 보다 효율적인 예산 분배가 가능.
- **정책 개선**: 예산 낭비를 줄여 출산장려금, 지역 교육 등 실질적 효과가 큰 정책에 자원 재배치를 하여 지역 발전의 균형을 맞추도록 유도.
- **지속 가능성**: 연도별·지자체별 예산 분석 결과를 축적하며 예측 모델의 정확도 향상 가능
