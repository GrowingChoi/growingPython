## Agent

## Agent 정리 요약

### Agent 개요  
- **Agent란**  
  - 대형 언어 모델과 다양한 도구를 결합하여 복잡한 작업을 처리할 수 있도록 설계된 사용자 요청 처리 인공지능 시스템  
  - 주어진 목표를 달성하기 위해 환경과 상호작용하며 의사결정을 내리고 행동하는 자율적 개체  
  - 사람의 개입을 최소화하고 LLM이 자율적으로 의사결정을 수행  

- **Agent의 특징**  
  - **자율성**: 사전 정의된 규칙 없이 스스로 결정하고 행동 가능  
  - **목표 지향성**: 특정 작업이나 목표 달성을 위한 구조를 가짐  
  - **도구 활용**: 다양한 도구나 API를 사용하여 작업 수행  

---

### Agent 기본 동작 원리  
- **ReAct (Reasoning and Acting)**  
  - 추론과 행동을 결합한 문제 해결 방식  
  - 동작 흐름:  
    1. 사용자 입력 수신  
    2. 추론을 통해 문제 분석 및 해결 계획 수립  
    3. 계획에 따라 도구 실행  
    4. 결과 평가 및 추가 도구 실행 여부 결정  
    5. 최종 결과 반환  

---

### Tool  
- **Tool이란**  
  - 하나의 기능을 처리하는 함수 단위의 구성 요소  

- **Tool Calling 개요**  
  - LLM이 외부 도구나 API를 호출해 작업을 처리할 수 있게 하는 기능  
  - 동작 방식:  
    1. 사용할 도구를 LLM에 바인딩  
    2. 사용자 질의 수신  
    3. LLM이 필요한 도구를 결정 및 호출  
    4. 도구 실행 결과를 기반으로 후속 작업 수행  

  - 예시:  
    - 수학 계산 시 `계산 도구` 호출  
    - 최신 정보 요청 시 `검색 도구` 또는 `DB 도구` 호출  

  - 특징:  
    - LLM은 도구를 통해 자체 지식의 한계를 넘어선 정보를 제공 가능  
    - OpenAI, Anthropic 등에서 지원하는 기능이며, 해당 기능을 지원하는 모델에서만 사용 가능  
    - LangChain은 다양한 도구 호출 방식을 표준화하고 일관된 인터페이스로 제공  
    - 검색 등 다양한 LangChain 내장 도구 제공

## 🔍 학습 요약

- **FACTS - 오늘배운 내용**  
  Agent는 LLM과 도구를 결합하여 자율적으로 문제를 해결하는 구조이며, ReAct 방식으로 동작하며 LangChain 기반으로 다양한 도구를 호출할 수 있다는 것을 배웠다.

- **FEELING - 학습 후 느낀점**  
  LLM이 단독으로 작동하는 한계를 도구 호출을 통해 확장할 수 있다는 점이 매우 흥미로웠고, 실제 서비스로 발전시킬 수 있겠다는 가능성을 느꼈다.

- **FINDINGS - 내가 필요하다고 생각되는 점**  
  직접 다양한 도구를 LangChain에 바인딩하고 에이전트를 구성해보는 실습이 필요하다.

- **FUTURE - 앞으로 어떤 방향으로 노력하고 공부해야할지**  
  LangChain에서 제공하는 Tool 인터페이스와 Agent 실행 구조를 실습하고, 추후에는 사용자 맞춤형 에이전트를 설계할 수 있는 수준으로 심화 학습할 계획이다.
